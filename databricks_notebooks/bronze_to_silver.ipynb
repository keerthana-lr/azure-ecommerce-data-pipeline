```python
storage_account_name = "stecommercepipeline"
bronze_path = f"abfss://bronze@{storage_account_name}.dfs.core.windows.net/"

display(dbutils.fs.ls(bronze_path))
```

```python
silver_path = f"abfss://silver@{storage_account_name}.dfs.core.windows.net/"

# Orders
orders_df = spark.read.csv(bronze_path + "olist_orders_dataset.csv", header=True, inferSchema=True)
orders_clean = orders_df.dropDuplicates(["order_id"]).na.drop(subset=["order_id", "customer_id"])
orders_clean.write.format("delta").mode("overwrite").save(silver_path + "orders")
print("Orders:", orders_clean.count())
```

```python
# Customers
customers_df = spark.read.csv(bronze_path + "olist_customers_dataset.csv", header=True, inferSchema=True)
customers_clean = customers_df.dropDuplicates(["customer_id"]).na.drop(subset=["customer_id"])
customers_clean.write.format("delta").mode("overwrite").save(silver_path + "customers")
print("Customers:", customers_clean.count())
```

```python
# Order Items
items_df = spark.read.csv(bronze_path + "olist_order_items_dataset.csv", header=True, inferSchema=True)
items_clean = items_df.dropDuplicates().na.drop(subset=["order_id", "product_id"])
items_clean.write.format("delta").mode("overwrite").save(silver_path + "order_items")
print("Order Items:", items_clean.count())
```

```python
# Payments
payments_df = spark.read.csv(bronze_path + "olist_order_payments_dataset.csv", header=True, inferSchema=True)
payments_clean = payments_df.dropDuplicates().na.drop(subset=["order_id"])
payments_clean.write.format("delta").mode("overwrite").save(silver_path + "payments")
print("Payments:", payments_clean.count())
```

```python
# Reviews
reviews_df = spark.read.csv(bronze_path + "olist_order_reviews_dataset.csv", header=True, inferSchema=True)
reviews_clean = reviews_df.dropDuplicates(["review_id"]).na.drop(subset=["order_id"])
reviews_clean.write.format("delta").mode("overwrite").save(silver_path + "reviews")
print("Reviews:", reviews_clean.count())
```

```python
# Products
products_df = spark.read.csv(bronze_path + "olist_products_dataset.csv", header=True, inferSchema=True)
products_clean = products_df.dropDuplicates(["product_id"]).na.drop(subset=["product_id"])
products_clean.write.format("delta").mode("overwrite").save(silver_path + "products")
print("Products:", products_clean.count())
```

```python
# Sellers
sellers_df = spark.read.csv(bronze_path + "olist_sellers_dataset.csv", header=True, inferSchema=True)
sellers_clean = sellers_df.dropDuplicates(["seller_id"]).na.drop(subset=["seller_id"])
sellers_clean.write.format("delta").mode("overwrite").save(silver_path + "sellers")
print("Sellers:", sellers_clean.count())
```
